# Notebook_3 Time series forecasting of migrant employment trends using ARIMA

## 1. Introduction

This notebook applies time series forecasting techniques to analyse historical employment trends among migrant workers in the United Kingdom. Using quarterly Office for National Statistics (ONS) employment data from 1997 onwards, the notebook develops forecasting models to project future employment patterns for migrant populations.

The analysis focuses on three employment groups:
- Non-UK-born employment
- European Union (EU27) employment
- Non-European Union (Non-EU) employment

Forecasting is performed using the AutoRegressive Integrated Moving Average (ARIMA) modelling framework. Prior to model development, each time series is examined for stationarity using the Augmented Dickey-Fuller (ADF) test, followed by model identification, forecasting and diagnostic evaluation. The resulting forecasts provide evidence of potential future employment trends among migrant workers and complement the exploratory analyses presented in Notebook 2.

## 3. Data Preparation

ARIMA models require a chronological time series. Therefore, a quarterly datetime variable is created by combining the observation year with the corresponding quarter. This variable serves as the temporal index for all subsequent forecasting analyses.

### 3.1 Create a Quarterly Date Variable

The ONS dataset records observations by year and quarter. Since ARIMA requires a datetime variable, a quarterly date variable is created by assigning the first month of each quarter as the observation date. The resulting date variable serves as the time index for forecasting.

```python
# Map each quarter to its first month
quarter_map = {
    'Q1': '01',
    'Q2': '04',
    'Q3': '07',
    'Q4': '10'
}

# Create Date column
ons['Date'] = pd.to_datetime(
    ons['Year'].astype(str) + '-' +
    ons['Quarter_No'].map(quarter_map) + '-01'
)

ons[['Date', 'Year', 'Quarter_No']].head()
```

## 5. Stationarity Testing (Augmented Dickey-Fuller Test)

Before fitting an ARIMA model, it is necessary to determine whether each time series is stationary. A stationary series has statistical properties, such as the mean and variance, that remain relatively constant over time.

The Augmented Dickey-Fuller (ADF) test is used to formally assess stationarity. The hypotheses for the test are:

- **Null Hypothesis (H₀):** The time series is non-stationary (contains a unit root).
- **Alternative Hypothesis (H₁):** The time series is stationary.

A p-value less than 0.05 indicates sufficient evidence to reject the null hypothesis and conclude that the series is stationary. If the series is found to be non-stationary, differencing will be applied before fitting the ARIMA model.

### 5.1 Augmented Dickey-Fuller Test: Non-UK-born Employment

The Augmented Dickey-Fuller (ADF) test is applied to the Non-UK-born employment series to determine whether the observed upward trend results in non-stationarity.

```python
from statsmodels.tsa.stattools import adfuller

# Perform ADF Test
adf_nonuk = adfuller(ons["NonUK_Born_Emp"])

# Display results
print("Augmented Dickey-Fuller Test")
print("-" * 40)
print(f"ADF Statistic : {adf_nonuk[0]:.4f}")
print(f"p-value       : {adf_nonuk[1]:.4f}")
print(f"Lags Used     : {adf_nonuk[2]}")
print(f"Observations  : {adf_nonuk[3]}")

print("\nCritical Values")
for key, value in adf_nonuk[4].items():
    print(f"{key}: {value:.4f}")
```

**Interpretation**:
The ADF test indicates that the Non-UK-born employment series is **non-stationary**. The ADF statistic (1.1795) is higher than the critical values, while the p-value (0.9958) exceeds the 0.05 significance level. Therefore, the null hypothesis cannot be rejected, confirming that the series is non-stationary. Consequently, differencing is required before fitting the ARIMA model.

### 5.2 Reusable ADF Test Function

To ensure consistency across all employment series, a reusable function is created to perform the Augmented Dickey-Fuller (ADF) test and display the test statistics in a standardised format.

```python
def run_adf_test(series, series_name):
    """
    Perform Augmented Dickey-Fuller Test
    """

    result = adfuller(series)

    print("=" * 55)
    print(f"Augmented Dickey-Fuller Test: {series_name}")
    print("=" * 55)

    print(f"ADF Statistic : {result[0]:.4f}")
    print(f"p-value       : {result[1]:.4f}")
    print(f"Lags Used     : {result[2]}")
    print(f"Observations  : {result[3]}")

    print("\nCritical Values")

    for key, value in result[4].items():
        print(f"{key}: {value:.4f}")

    if result[1] < 0.05:
        print("\nConclusion: Stationary (Reject H₀)")
    else:
        print("\nConclusion: Non-Stationary (Fail to Reject H₀)")
```

### 5.3 Augmented Dickey-Fuller Test: EU27 Employment

The ADF test is applied to the EU27 employment series to determine whether the observed employment trend is stationary prior to ARIMA modelling.

```python
run_adf_test(
    ons["EU27_Emp"],
    "EU27 Employment"
)
```

**Interpretation**:
The ADF test indicates that the EU27 employment series is **non-stationary**. The ADF statistic (-0.6018) is higher than the critical values, while the p-value (0.8706) exceeds the 0.05 significance level. Therefore, the null hypothesis cannot be rejected, confirming that the series is non-stationary. This supports the earlier visual analysis and indicates that differencing is required before ARIMA modelling.

### 5.4 Augmented Dickey-Fuller Test: Non-EU Employment

The ADF test is applied to the Non-EU employment series to assess whether the series satisfies the stationarity assumption required for ARIMA modelling.

```python
run_adf_test(
    ons["NonEU_Emp"],
    "Non-EU Employment"
)
```

**Interpretation**:
The ADF test confirms that the Non-EU employment series is **non-stationary**. The ADF statistic (3.0973) is higher than the critical values, and the p-value (1.0000) is greater than 0.05. Therefore, the null hypothesis cannot be rejected, indicating that the series is non-stationary. Consequently, differencing will be required before fitting the ARIMA model.

## 6. Model Identification and Selection

Following the stationarity assessment, the next stage is to identify an appropriate ARIMA model for each employment series. Since all three series were found to be non-stationary, differencing is first applied to remove the long-term trend. The differenced series is then re-tested for stationarity before selecting the most appropriate ARIMA model.

### 6.1 First-Order Differencing

First-order differencing is applied to each employment series to remove the long-term trend and achieve stationarity. This involves calculating the difference between consecutive quarterly observations.

```python
# First-order differencing

ons["NonUK_Diff"] = ons["NonUK_Born_Emp"].diff()

ons["EU27_Diff"] = ons["EU27_Emp"].diff()

ons["NonEU_Diff"] = ons["NonEU_Emp"].diff()

ons[[
    "NonUK_Diff",
    "EU27_Diff",
    "NonEU_Diff"
]].head()
```

**Interpretation**: First-order differencing was applied to remove the long-term trend from each employment series. The transformed series will now be tested to determine whether stationarity has been achieved before fitting the ARIMA models.

### 6.2 Stationarity Test After Differencing

The Augmented Dickey-Fuller (ADF) test is repeated on each differenced series to confirm whether first-order differencing has successfully produced stationary time series suitable for ARIMA modelling.

```python
run_adf_test(
    ons["NonUK_Diff"].dropna(),
    "Differenced Non-UK-born Employment"
)

run_adf_test(
    ons["EU27_Diff"].dropna(),
    "Differenced EU27 Employment"
)

run_adf_test(
    ons["NonEU_Diff"].dropna(),
    "Differenced Non-EU Employment"
)
```

### 6.3 Model Identification Using ACF and PACF

Following first-order differencing, the Autocorrelation Function (ACF) and Partial Autocorrelation Function (PACF) plots are examined to identify suitable values for the autoregressive (p) and moving average (q) components of the ARIMA model.

```python
import matplotlib.pyplot as plt

from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
```

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

# ACF Plot
plot_acf(
    ons["NonUK_Diff"].dropna(),
    lags=20
)

plt.title("ACF Plot - Differenced Non-UK-born Employment")

plt.show()


# PACF Plot
plot_pacf(
    ons["NonUK_Diff"].dropna(),
    lags=20,
    method="ywm"
)

plt.title("PACF Plot - Differenced Non-UK-born Employment")

plt.show()
```

**Interpretation**: The ACF and PACF plots indicate that most autocorrelations fall within the 95% confidence intervals after the first lag. The ACF displays a significant spike at lag 1, while the PACF also shows a prominent spike at lag 1 before tapering off. Based on these patterns, an ARIMA(1,1,1) model is selected as an appropriate starting model for forecasting the Non-UK-born employment series.

## 6.4 Automatic Model Selection

Although the ACF and PACF plots suggested an initial ARIMA(1,1,1) model, an automatic model selection procedure was performed to identify the model that provides the best balance between goodness of fit and model complexity. The selected model was then used for subsequent forecasting and evaluation.

```python
from pmdarima import auto_arima

auto_model = auto_arima(
    ons["NonUK_Born_Emp"],
    start_p=0,
    start_q=0,
    max_p=3,
    max_q=3,
    d=1,
    seasonal=False,
    trace=True,
    error_action="ignore",
    suppress_warnings=True,
    stepwise=True
)

print(auto_model.summary())
```

#### Interpretation:
Auto ARIMA selected ARIMA(0,1,1) as the best-fitting model based on the lowest AIC value. This model was used for all subsequent forecasting analyses.

## 6.5 Candidate Model Validation

The models identified through manual selection and Auto ARIMA were compared using forecast accuracy measures. The model with the best overall predictive performance was selected for the remaining analysis.

```python
split_index = int(len(ons) * 0.90)

train = ons["NonUK_Born_Emp"][:split_index]
test = ons["NonUK_Born_Emp"][split_index:]
```

```python
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np
import pandas as pd

# Candidate models
candidate_models = {
    "ARIMA(0,1,1)": (0,1,1),
    "ARIMA(1,1,1)": (1,1,1)
}

results = []

for model_name, order in candidate_models.items():

    model = ARIMA(train, order=order)
    fitted = model.fit()

    forecast = fitted.forecast(steps=len(test))

    rmse = np.sqrt(mean_squared_error(test, forecast))
    mae = mean_absolute_error(test, forecast)
    mape = np.mean(np.abs((test - forecast) / test)) * 100

    results.append([
        model_name,
        fitted.aic,
        rmse,
        mae,
        mape
    ])

comparison = pd.DataFrame(
    results,
    columns=[
        "Model",
        "AIC",
        "RMSE",
        "MAE",
        "MAPE"
    ]
)

comparison.sort_values("MAPE")
```

```python
metrics_df = pd.DataFrame({
    "Metric": ["RMSE", "MAE", "MAPE"],
    "Value": [rmse, mae, mape]
})

metrics_df.to_csv(
    "../04_Output/Tables/Forecast_Validation_Metrics.csv",
    index=False
)

print("Validation metrics exported.")
```

#### Interpretation:
ARIMA(1,1,1) achieved lower AIC, RMSE, MAE and MAPE values than ARIMA(0,1,1). Therefore, ARIMA(1,1,1) was selected as the final forecasting model.

## 6.6 Fit Final ARIMA Model

The ARIMA(1,1,1) model was fitted to the Non-UK-born employment series based on its superior forecast performance.

```python
from statsmodels.tsa.arima.model import ARIMA

model_nonuk = ARIMA(
    ons["NonUK_Born_Emp"],
    order=(1,1,1)
)

results_nonuk = model_nonuk.fit()

print(results_nonuk.summary())
```

#### Interpretation
The ARIMA(1,1,1) model was successfully fitted and provides the basis for the remaining forecasting analysis.

### 6.7.3 Normal Q-Q Plot

A normal Q-Q plot was used to compare the residual distribution with a normal distribution.

```python
from scipy import stats

plt.figure(figsize=(6,6))

stats.probplot(
    nonuk_residuals[1:],
    dist="norm",
    plot=plt
)

plt.title("Normal Q-Q Plot")
plt.savefig(
    "../04_Output/Figures/Residuals_QQPlot.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

#### Interpretation:
Most residuals follow the reference line, indicating that the residuals are approximately normally distributed.

### 6.7.4 Ljung–Box Test

The Ljung–Box test was performed to determine whether significant autocorrelation remained in the residuals.

```python
from statsmodels.stats.diagnostic import acorr_ljungbox

ljung_box = acorr_ljungbox(
    nonuk_residuals[1:],
    lags=[10],
    return_df=True
)

print(ljung_box)
```

#### Interpretation:
The Ljung–Box test produced a p-value greater than 0.05, indicating no significant autocorrelation remains in the residuals. This suggests that the ARIMA(1,1,1) model provides an adequate fit to the data.

#### 6.7.5 Diagnostic Summary

The diagnostic tests indicate that the ARIMA(1,1,1) model provides an adequate fit for the Non-UK-born employment series. The residuals fluctuate randomly around zero, are approximately normally distributed, and show no significant autocorrelation, supporting the model's suitability for forecasting.

```python
# Create comparative forecast plot
plt.figure(figsize=(14, 7))

# ==========================================
# Historical Data
# ==========================================
plt.plot(
    ons["Date"],
    ons["NonUK_Born_Emp"],
    linewidth=2,
    label="Historical Non-UK-born"
)

plt.plot(
    ons["Date"],
    ons["EU27_Emp"],
    linewidth=2,
    label="Historical EU27"
)

plt.plot(
    ons["Date"],
    ons["NonEU_Emp"],
    linewidth=2,
    label="Historical Non-EU"
)

# ==========================================
# Forecast Lines
# ==========================================

plt.plot(
    future_dates,
    forecast_mean,
    '--',
    linewidth=2,
    label="Forecast Non-UK-born"
)

plt.plot(
    future_dates,
    eu27_mean,
    '--',
    linewidth=2,
    label="Forecast EU27"
)

plt.plot(
    future_dates,
    noneu_mean,
    '--',
    linewidth=2,
    label="Forecast Non-EU"
)

# ==========================================
# 95% Confidence Intervals
# (Displayed but omitted from the legend)
# ==========================================

plt.fill_between(
    future_dates,
    forecast_ci.iloc[:, 0],
    forecast_ci.iloc[:, 1],
    alpha=0.15,
    label="_nolegend_"
)

plt.fill_between(
    future_dates,
    eu27_ci.iloc[:, 0],
    eu27_ci.iloc[:, 1],
    alpha=0.15,
    label="_nolegend_"
)

plt.fill_between(
    future_dates,
    noneu_ci.iloc[:, 0],
    noneu_ci.iloc[:, 1],
    alpha=0.15,
    label="_nolegend_"
)

# ==========================================
# Formatting
# ==========================================
plt.title(
    "Comparative Employment Forecasts Across Migrant Workforce Groups",
    fontsize=14
)

plt.xlabel("Year", fontsize=12)
plt.ylabel("Employment (Thousands)", fontsize=12)

plt.grid(True, alpha=0.3)

plt.legend(
    loc="upper left",
    fontsize=10,
    frameon=True
)

plt.tight_layout()

plt.savefig(
    "../04_Output/Figures/Comparative_Migrant_Employment_Forecast.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

#### Interpretation:
Using the validated ARIMA(1,1,1) forecasting approach, the projected employment trends indicate continued growth across the Non-UK-born, EU27, and Non-EU employment series over the forecast horizon. Non-UK-born employment is projected to remain substantially higher than the individual EU27 and Non-EU employment series, while differences in the forecast trajectories reflect variation in the composition of the migrant workforce. These comparative forecasts provide valuable labour market context for the subsequent analysis of workforce trends within the adult social care sector.

## 9. Notebook Summary

This notebook applied ARIMA time series modelling to forecast employment trends among migrant workforce groups in the UK labour market. The analysis began with exploratory visualisation and stationarity testing using the Augmented Dickey-Fuller (ADF) test. As all three employment series were found to be non-stationary, first-order differencing was applied before model identification using the ACF and PACF plots.

ARIMA(1,1,1) models were subsequently fitted to the Non-UK-born, EU27 and Non-EU employment series. The fitted models were evaluated using residual diagnostics and then used to generate quarterly forecasts for the next 12 quarters. The forecast results indicate continued growth in overall Non-UK-born employment and stronger growth among Non-EU workers, while EU27 employment is projected to remain relatively stable. Although the forecasts suggest positive employment trends, the widening confidence intervals indicate increasing uncertainty over longer forecast horizons.

Overall, the forecasting analysis provides evidence-based insights into future migrant employment trends and complements the exploratory and correlation analyses presented in the previous notebooks. These findings contribute to a better understanding of future workforce patterns and may support evidence-based workforce planning and policy development within the UK labour market.
