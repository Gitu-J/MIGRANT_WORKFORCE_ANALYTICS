## Notebook 4: Skills for Care Workforce Forecasting

#### Introduction

This notebook extends the previous analyses by forecasting key workforce indicators from the Skills for Care dataset. While Notebook 3 forecasted broader migrant employment trends using ONS data, this notebook focuses directly on workforce challenges within the adult social care sector.

The analysis forecasts:

Vacancy Rate (Vacancy_Rt)
Turnover Rate (Turnover_Rt)

These indicators are closely linked to recruitment difficulties, workforce shortages, and staff retention, making them particularly relevant to the study's aim of understanding workforce challenges and retention patterns in the UK adult social care sector.

#### 1. Import Libraries

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.arima.model import ARIMA

from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error

import warnings
warnings.filterwarnings("ignore")
```

#### 2. Load and Prepare Skills for Care Data
#### 2.1 Load Dataset

```python
sfc = pd.read_excel(
    "../02_Clean_Data/Skills_for_Care/Skills_for_Care_Cleaned_Dataset.xlsx"
)
```

#### 2.2 Inspect Variables

```python
sfc.columns
sfc.head()
```

#### 2.3 Select Forecast Variables

```python
forecast_data = sfc[
    ['Year', 'Vacancy_Rt', 'Turnover_Rt']
].copy()

forecast_data.head()
```

#### 2.4 Convert Year to Datetime

```python
forecast_data['Year'] = pd.to_datetime(
    forecast_data['Year'].astype(str)
)

forecast_data.set_index(
    'Year',
    inplace=True
)
```

#### 3. Stationarity Assessment
#### 3.1 ADF Test Function

```python
def adf_test(series, variable_name):

    result = adfuller(series.dropna())

    print(f"\nADF Test for {variable_name}")
    print(f"ADF Statistic: {result[0]:.4f}")
    print(f"P-value: {result[1]:.4f}")

    if result[1] < 0.05:
        print("Series appears stationary")
    else:
        print("Series appears non-stationary")
```

#### 3.2 Vacancy Rate

```python
adf_test(
    forecast_data['Vacancy_Rt'],
    'Vacancy Rate'
)
```

#### 3.3 Turnover Rate

```python
adf_test(
    forecast_data['Turnover_Rt'],
    'Turnover Rate'
)
```

#### Interpretation

Because only ten annual observations are available, ADF results should be interpreted cautiously. The test is used as a diagnostic indicator rather than definitive evidence of stationarity.

#### 4. Differencing
#### 4.1 Vacancy Rate

```python
vacancy_diff = (
    forecast_data['Vacancy_Rt']
    .diff()
    .dropna()
)
```

#### 4.2 Turnover Rate

```python
turnover_diff = (
    forecast_data['Turnover_Rt']
    .diff()
    .dropna()
)
```

#### 4.3 Re-test Stationarity

```python
adf_test(
    vacancy_diff,
    'Differenced Vacancy Rate'
)
```

```python
adf_test(
    turnover_diff,
    'Differenced Turnover Rate'
)
```

#### 5. ACF and PACF Analysis
#### 5.1 Vacancy Rate

```python
fig, axes = plt.subplots(
    1,
    2,
    figsize=(12,4)
)

plot_acf(
    vacancy_diff,
    ax=axes[0]
)

plot_pacf(
    vacancy_diff,
    ax=axes[1]
)

axes[0].set_title(
    'Vacancy Rate ACF'
)

axes[1].set_title(
    'Vacancy Rate PACF'
)

plt.tight_layout()

plt.savefig(
    "../04_Output/Figures/Vacancy_ACF_PACF.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

#### 6. Train-Test Split

The final two years are reserved for validation.

#### 6.1 Vacancy Rate

```python
vacancy_train = (
    forecast_data['Vacancy_Rt'][:-2]
)

vacancy_test = (
    forecast_data['Vacancy_Rt'][-2:]
)
```

#### 6.2 Turnover Rate

```python
turnover_train = (
    forecast_data['Turnover_Rt'][:-2]
)

turnover_test = (
    forecast_data['Turnover_Rt'][-2:]
)
```

#### 7. ARIMA Model Selection

Three candidate models are evaluated:

ARIMA(1,1,0)
ARIMA(0,1,1)
ARIMA(1,1,1)

#### 7.1 Vacancy Rate Models

```python
vacancy_models = {

    "ARIMA(1,1,0)":
        ARIMA(
            vacancy_train,
            order=(1,1,0)
        ).fit(),

    "ARIMA(0,1,1)":
        ARIMA(
            vacancy_train,
            order=(0,1,1)
        ).fit(),

    "ARIMA(1,1,1)":
        ARIMA(
            vacancy_train,
            order=(1,1,1)
        ).fit()
}
```

#### Compare AIC and BIC

```python
vacancy_model_comparison = pd.DataFrame({

    'Model':
        vacancy_models.keys(),

    'AIC':
        [m.aic for m in vacancy_models.values()],

    'BIC':
        [m.bic for m in vacancy_models.values()]
})
vacancy_model_comparison.to_csv(
    "../04_Output/Tables/Vacancy_Model_Comparison.csv",
    index=False
)
vacancy_model_comparison
```

#### 7.2 Turnover Rate Model Selection

```python
turnover_models = {

    "ARIMA(1,1,0)":
        ARIMA(
            turnover_train,
            order=(1,1,0)
        ).fit(),

    "ARIMA(0,1,1)":
        ARIMA(
            turnover_train,
            order=(0,1,1)
        ).fit(),

    "ARIMA(1,1,1)":
        ARIMA(
            turnover_train,
            order=(1,1,1)
        ).fit()
}
```

#### Compare AIC and BIC

```python
turnover_model_comparison = pd.DataFrame({

    'Model':
        turnover_models.keys(),

    'AIC':
        [m.aic for m in turnover_models.values()],

    'BIC':
        [m.bic for m in turnover_models.values()]
})
turnover_model_comparison.to_csv(
    "../04_Output/Tables/Turnover_Model_Comparison.csv",
    index=False
)
turnover_model_comparison
```

#### 8. Forecast Validation
#### 8.1 Vacancy Rate Validation Model

```python
best_vacancy_model = ARIMA(
    vacancy_train,
    order=(0,1,1)
).fit()

vacancy_forecast = best_vacancy_model.forecast(
    steps=len(vacancy_test)
)
```

#### 8.2 Turnover Rate Validation Model

```python
best_turnover_model = ARIMA(
    turnover_train,
    order=(0,1,1)
).fit()

turnover_forecast = best_turnover_model.forecast(
    steps=len(turnover_test)
)
```

#### Interpretation
The selected ARIMA(0,1,1) model is fitted using the training dataset and used to forecast the validation period. Forecast performance is subsequently assessed using RMSE, MAE, and MAPE.

#### 9. Forecast Accuracy Assessment
#### 9.1 Vacancy Rate Accuracy Metrics

```python
vacancy_rmse = np.sqrt(
    mean_squared_error(
        vacancy_test,
        vacancy_forecast
    )
)

vacancy_mae = mean_absolute_error(
    vacancy_test,
    vacancy_forecast
)

vacancy_mape = np.mean(
    np.abs(
        (vacancy_test - vacancy_forecast)
        / vacancy_test
    )
) * 100

print(f"RMSE: {vacancy_rmse:.4f}")
print(f"MAE: {vacancy_mae:.4f}")
print(f"MAPE: {vacancy_mape:.2f}%")
```

#### 9.2 Turnover Rate Accuracy Metrics

```python
turnover_rmse = np.sqrt(
    mean_squared_error(
        turnover_test,
        turnover_forecast
    )
)

turnover_mae = mean_absolute_error(
    turnover_test,
    turnover_forecast
)

turnover_mape = np.mean(
    np.abs(
        (turnover_test - turnover_forecast)
        / turnover_test
    )
) * 100

print(f"RMSE: {turnover_rmse:.4f}")
print(f"MAE: {turnover_mae:.4f}")
print(f"MAPE: {turnover_mape:.2f}%")
```

#### 9.3 Validation Metrics Summary

```python
validation_metrics = pd.DataFrame({

    'Variable': [
        'Vacancy Rate',
        'Turnover Rate'
    ],

    'RMSE': [
        vacancy_rmse,
        turnover_rmse
    ],

    'MAE': [
        vacancy_mae,
        turnover_mae
    ],

    'MAPE': [
        vacancy_mape,
        turnover_mape
    ]
})

validation_metrics
```

#### 9.4 Export Validation Metrics

```python
validation_metrics.to_csv(
    "../04_Output/Tables/SkillsForCare_Forecast_Validation_Metrics.csv",
    index=False
)

print("Validation metrics exported successfully.")
```

#### 10. Final Model Fitting
#### 10.1 Vacancy Rate Final Model

```python
vacancy_final = ARIMA(
    forecast_data['Vacancy_Rt'],
    order=(0,1,1)
).fit()

print(vacancy_final.summary())
```

#### 10.2 Turnover Rate Final Model

```python
turnover_final = ARIMA(
    forecast_data['Turnover_Rt'],
    order=(0,1,1)
).fit()

print(turnover_final.summary())
```

#### 11. Future Workforce Forecasts (2027–2029)
#### 11.1 Generate Future Forecasts

A three-year forecast horizon is selected to provide short-term workforce projections while minimising uncertainty associated with long-term forecasting.

```python
forecast_horizon = 3

future_years = [
    2027,
    2028,
    2029
]

vacancy_future = vacancy_final.forecast(
    steps=forecast_horizon
)

turnover_future = turnover_final.forecast(
    steps=forecast_horizon
)
```

#### 11.2 Forecast Results Table

```python
forecast_results = pd.DataFrame({

    'Year': future_years,

    'Vacancy_Rate_Forecast':
        vacancy_future.values,

    'Turnover_Rate_Forecast':
        turnover_future.values

})
forecast_results.to_csv(
    "../04_Output/Tables/SkillsForCare_Workforce_Forecast.csv",
    index=False
)

print("Forecast results exported successfully.")

forecast_results
```

#### Interpretation

The forecast results provide projected vacancy rates and turnover rates for the period 2027–2029. These projections indicate the likely direction of future workforce pressures within the adult social care sector. Increasing vacancy rates would suggest worsening recruitment challenges, while increasing turnover rates would indicate growing staff retention difficulties. Conversely, declining rates would suggest improvements in workforce stability.

#### 12. Forecast Visualisation
#### 12.1 Vacancy Rate Forecast

```python
plt.figure(figsize=(10,5))

combined_years = (
    list(forecast_data.index.year)
    + future_years
)

combined_values = (
    list(forecast_data['Vacancy_Rt'])
    + list(vacancy_future)
)

plt.plot(
    combined_years,
    combined_values,
    marker='o',
    linewidth=2
)

plt.axvline(
    x=2026,
    linestyle='--',
    label='Forecast Start'
)

plt.title(
    'Adult Social Care Vacancy Rate Forecast (2017–2029)'
)

plt.xlabel('Year')
plt.ylabel('Vacancy Rate')

plt.legend()
plt.grid(True)
plt.savefig(
    "../04_Output/Figures/Vacancy_Rate_Forecast.png",
    dpi=300,
    bbox_inches='tight'
)
plt.show()
```

#### 12.2 Turnover Rate Forecast

```python
plt.figure(figsize=(10,5))

combined_years = (
    list(forecast_data.index.year)
    + future_years
)

combined_values = (
    list(forecast_data['Turnover_Rt'])
    + list(turnover_future)
)

plt.plot(
    combined_years,
    combined_values,
    marker='o',
    linewidth=2,
    color='tab:orange',
    label='Turnover Rate'
)

plt.axvline(
    x=2026,
    linestyle='--',
    label='Forecast Start'
)

plt.title(
    'Adult Social Care Turnover Rate Forecast (2017–2029)'
)

plt.xlabel('Year')
plt.ylabel('Turnover Rate (%)')

plt.legend()
plt.grid(True)

plt.savefig(
    "../04_Output/Figures/Turnover_Rate_Forecast.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

#### 12.3 Comparative Forecast Visualisation

```python
plt.figure(figsize=(10,5))

vacancy_combined = (
    list(forecast_data['Vacancy_Rt'])
    + list(vacancy_future)
)

turnover_combined = (
    list(forecast_data['Turnover_Rt'])
    + list(turnover_future)
)

combined_years = (
    list(forecast_data.index.year)
    + future_years
)

plt.plot(
    combined_years,
    vacancy_combined,
    marker='o',
    linewidth=2,
    label='Vacancy Rate'
)

plt.plot(
    combined_years,
    turnover_combined,
    marker='o',
    linewidth=2,
    label='Turnover Rate'
)

plt.axvline(
    x=2026,
    linestyle='--',
    label='Forecast Start'
)

plt.title(
    'Adult Social Care Workforce Forecast Trends (2017–2029)'
)

plt.xlabel('Year')
plt.ylabel('Rate (%)')

plt.legend()
plt.grid(True)

plt.savefig(
    "../04_Output/Figures/Comparative_Workforce_Forecast.png",
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```

### 13. Summary
This notebook forecasted vacancy and turnover rates in the UK adult social care sector using ARIMA models. The results indicate relatively stable workforce trends from 2027 to 2029 and provide insights into future workforce sustainability and retention patterns within the sector.
